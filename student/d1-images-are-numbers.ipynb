`ndimage.label` finds every blob of touching `True` pixels and gives each blob
its own number: 1, 2, 3 and so on, with 0 for the background. It also tells you
how many blobs it found.

```python
labels, n_blobs = ndimage.label(nuclei_mask)
print("blobs found:", n_blobs)
show(labels, "each blob in its own colour", cmap="nipy_spectral")
```

Some of those blobs are a few pixels of noise, not nuclei. `np.bincount` counts
how many pixels carry each label, so `blob_sizes[k]` is the size of blob `k`
(and `blob_sizes[0]` is the background).

```python
blob_sizes = np.bincount(labels.ravel())
print("sizes of the first ten blobs:", blob_sizes[1:11])
```
